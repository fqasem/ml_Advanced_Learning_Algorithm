```python
from datetime import datetime, timezone

# Recommended: Get current time in UTC
utc_now = datetime.now(timezone.utc)
print(utc_now)
print(utc_now.isoformat())  
# Output: 2026-06-18T15:23:01.123456+00:00
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import current_timestamp, to_utc_timestamp, date_format

# Initialize Spark session
spark = SparkSession.builder.appName("UTC_Timestamp").getOrCreate()

# Create a single-row DataFrame to generate the timestamps
df = spark.range(1)

# Generate UTC time and its ISO format string
df_result = df.withColumn("utc_now", to_utc_timestamp(current_timestamp(), "UTC")) \
              .withColumn("utc_now_iso", date_format("utc_now", "yyyy-MM-dd'T'HH:mm:ss.SSSXXX"))

# Show the results
df_result.show(truncate=False)
```
